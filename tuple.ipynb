# change & remove elem

```python
tuple=('Pragati','Tejal','You','me','one','two')
```

```python
tuple[2]= 'Three'    # immutable (can't change)
```

```python
del tuple[3]    # can't delete specific elem
```

```python
del tuple   # delete entire tuple
```

# tuple membership

```python
tuple=('Pragati','Tejal','You','me','one','two')
```

```python
'one' in tuple
```

```python
'tejal' in tuple   # Tejal
```

```python
if 'You' in tuple:
    print('two is present in tuple')
else:
    print('two is not present in tuple')
```

```python
if 'tejal' in tuple:
    print('tejal is present in tuple')
else:
    print('tejal is not present in tuple')
```
