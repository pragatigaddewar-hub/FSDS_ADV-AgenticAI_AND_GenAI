slicing     [ : ]       

forward slicing  

```python
s              #   h  e  l  l  o
                   0  1  2  3  4 
```

```python
s[0:3]      #  0th index no. to last -->  (n-1)
```

```python
s[2:5]           # 2th index no. to    (n-1) =>  (5-1)
```

```python
s[ 0 : ]         #  0th to  last (whole) or length of str
```

```python
s[ : 3 ]         # starting  0th to (3-1) = 2th
```

```python
s[ : 7]      #   total index no. 0th to 6th  but  we take 7 means whole str is cover
```

```python
s[2:10]          #  index no. 2th to 6th  but  we take 10 means whole str is cover
```

```python
s[0:100]              #   total index no. 0th to 6th  but  we take 100 means whole str is cover    ( till -----)
```

```python
s[7]                           # there is no 7th index no.  
```

```python
s[0:300]           # in slicing this is possible
```

```python
s[-4:-2]                             # -4  to  (n-1) = (-4:(-3))
```
