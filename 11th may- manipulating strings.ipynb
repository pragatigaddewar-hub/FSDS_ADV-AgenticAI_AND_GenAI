# text with rjust() , ljust() , center    --> spaces

# rjust()

```python
'Hello'.rjust(10)             # After giving 10 spaces,word towards right 
```

```python
'meeeeeee'.rjust(66)    # word towards right
```

# ljust()

```python
'world'.ljust(30)              # word towards left, after word giving 30 spaces 
```

```python
'you'.ljust(5)            # word in left
```

```python
'morning'.center(11)            # 11 - 7 = 4  ,each side 2-2
```

```python
'hello'.rjust(20,'*')
```

```python
'hello'.ljust(40,'&')
```

```python
'hello'.rjust(30,'=')
```
