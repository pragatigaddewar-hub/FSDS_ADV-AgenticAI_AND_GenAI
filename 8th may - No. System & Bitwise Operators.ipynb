Bitwise Operators

1) complement (~)
2) bitwise and (&)
3) bitwise or ( | )
4) xor  (^)
5) left shift ( << )
6) right shift  ( >> )

1) Compliment (~)       # always store -ve value
                        #  -(n+1)   

```python
~15
```

```python
~26
```

                0  0  = 0

```python
25 ^ 30
```

5) left shift ( << )   # gain by 0 in last

```python
12 << 2         # gain by 2  zeroes infront of 12
                # 12 =>  1100   <<  2    = 1100 00  =   1     1      0      0     0      0
                                                    #  2^5   2^4    2^3    2^2   2^1     2^0

                                                   # =  32 + 16  
```

```python
13 << 1
```

6) right shift ( >> )       # reduce / cut last by

```python
12 >> 2                #  12 =>  1100     >>2 so last 2 numbers cut  
                       #         11   = 2^1 + 2^0    = 2 + 1
```

```python
13 >> 3        # last 3 numbers cut 
               # 13 => 1101  
                #    = 1    = 1 * 2^0    =   1       
```
