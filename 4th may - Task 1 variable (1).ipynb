1. storing and printing a value

```python
# storing the value 10 in a variable called 'x'

x=10                 

# printing the value of x

print(x)                  # output : 10
```

2. Using variable in expressions

```python
# assigning values to variables
a=5
b=8                      # adding two variables and storing the result in 'result'

result = a + b     
 
# printing the result

print(result)              # output : 13
                    
```

3. changing the value of a variable

```python
# initial value

score=50

print(score)        # output : 50  


# changing the value of the score

score=100

print(score)            #output : 100

```

4. concatenating strings

```python
# Assigning values to variables

first_name= "Pragati"
last_name= "Gaddewar"

# concatenating strings and storing in a new variable

full_name = first_name + " " + last_name

print(full_name)                              # output : Pragati Gaddewar
```

5. Using variables in a Calculation

```python
# assigning values to variable

length= 13
width=2

# calculating the area of a rectangle

area = length * width
print(area)      # output : 26
```
